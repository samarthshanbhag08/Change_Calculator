```python
cost = float(input("Cost: $"))
money = float(input("Money given: $"))
```

```python
change = float(money-cost)
change= round(change,2)
change = change*100
```

```python
if change>25:
    Quaters = change//25
    change %=25
if change>10:
    Dime = change//10
    change %=10
if change>5:
    Nickel = change//5
    change %=5
if change>1:
    Penny = change//1
    change %=1
print(f' Quaters:{Quaters} \n Dime:{Dime} \n Nickel:{Nickel} \n Penny:{Penny}')
```
