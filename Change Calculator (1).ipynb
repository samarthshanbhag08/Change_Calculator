```python
cost = float(input("Cost: $"))
money = float(input("Money given: $"))
change = float(money-cost)
change= round(change,2)
```

```python
change = change*100
Quarters = 0
Dime = 0
Nickel = 0
Penny = 0
```

```python
if change>=25:
    Quarters = change//25
    change %=25
if change>=10:
    Dime = change//10
    change %=10
if change>=5:
    Nickel = change//5
    change %=5
if change>=1:
    Penny = change//1
    change %=1
print(f' Quarters:{Quarters} \n Dime:{Dime} \n Nickel:{Nickel} \n Penny:{Penny}')
```
